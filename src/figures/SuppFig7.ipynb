In this notebook, I will make plots for Supplemental Figure 7.

# Setup

```python
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

thisseed = 231103

sc.settings.seed = thisseed

colFGID = '#1E88E5'
coliCD = '#55D276'
colcCD = '#FFC107'
colUC = '#D81B60'
colIBD = '#EB6D33'

colNM = '#6b6a6a'
colNOA = '#bfbfbf'
colFR = '#11d6cc'
colPR = '#ff8c00'
colNR = '#ff1e00'

colBG = '#C2C2C2'

sc.set_figure_params(dpi=100)

plt.rcParams.update({'font.weight': 'bold'})
plt.rcParams.update({'axes.labelweight': 'bold'})
plt.rcParams.update({'axes.titleweight': 'bold'})
```

# ColEpi

## Setup

```python
adata_colepi = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_colepi_annotated_251017.h5ad')
adata_colepi
```

```python
wsum_norm_colepi = pd.read_csv('../../predict/data/decoupler/il17/wsum_kegg/colepi__il17_scoring_wsum_aucell.csv')
wsum_norm_colepi.set_index('barcode', inplace=True)
wsum_norm_colepi.index.name = None
wsum_norm_colepi
```

```python
colnames = list(wsum_norm_colepi.columns)
colnames
```

```python
adata_colepi.obs[colnames] = wsum_norm_colepi
adata_colepi.obs
```

```python
colepi_fgid = adata_colepi.obs.diagSub.cat.reorder_categories(['iCD', 'cCD', 'UC', 'FGID']).sort_values().index.tolist()
colepi_ccd = adata_colepi.obs.diagSub.cat.reorder_categories(['FGID', 'iCD', 'UC', 'cCD']).sort_values().index.tolist()
colepi_uc = adata_colepi.obs.diagSub.cat.reorder_categories(['FGID', 'iCD', 'cCD', 'UC']).sort_values().index.tolist()
```

## UMAPs

```python
sc.pl.umap(adata_colepi[colepi_fgid][adata_colepi.obs.timepoint.isin(['T0'])], 
           color='diagSub', palette=[colFGID, colBG, colBG, colBG], 
           legend_loc='None', title='Colonic Epithelium (FGID)') 
```

```python
sc.pl.umap(adata_colepi[colepi_ccd][adata_colepi.obs.timepoint.isin(['T0'])], 
           color='diagSub', palette=[colBG, colBG, colcCD, colBG], 
           legend_loc='None', title='Colonic Epithelium (cCD)') 
```

```python
sc.pl.umap(adata_colepi[colepi_uc][adata_colepi.obs.timepoint.isin(['T0'])], 
           color='diagSub', palette=[colBG, colBG, colBG, colUC], 
           legend_loc='None', title='Colonic Epithelium (UC)') 
```

```python
sc.pl.umap(adata_colepi[adata_colepi.obs.timepoint.isin(['T0'])], 
           color=['il17af_antimicrobial'], 
           cmap='turbo', legend_loc='none', 
           vmin=0, vmax=5, ncols=2,
          title=['IL17 Antimicrobial Score'])
```

```python
sc.pl.umap(adata_colepi[adata_colepi.obs.timepoint.isin(['T0'])], 
           color=['il17af_chemokines'], 
           cmap='turbo', legend_loc='none', 
           vmin=0, vmax=5, ncols=2,
          title=['IL17 Chemokine Score'])
```
