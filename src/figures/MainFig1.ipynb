This notebook will create the UMAPs in Main Figure 1

# Setup

```python
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sc.settings.verbosity = 3

thisseed = 231103
np.random.seed(thisseed)
```

# Cell State UMAP

```python
sc.set_figure_params(dpi=200, dpi_save=200)

plt.rcParams.update({'font.weight': 'bold'})
plt.rcParams.update({'axes.labelweight': 'bold'})
plt.rcParams.update({'axes.titleweight': 'bold'})
```

## All Cells

```python
adata = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_allcells_annotated_251017.h5ad')
adata
```

```python
sc.pl.umap(adata, color='celltype', title='Colonic Cells')
```

## ColEpi

```python
adata_colepi = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_colepi_annotated_251017.h5ad')
adata_colepi
```
