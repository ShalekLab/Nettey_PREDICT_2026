# Setup

```python
import scanpy as sc
import numpy as np
import matplotlib.pyplot as plt
from random import sample
import pandas as pd

sc.settings.verbosity = 3

thisseed = 231103
np.random.seed(thisseed)
```

# Tfh Tph

```python
sc.set_figure_params(dpi=80)

plt.rcParams.update({'font.size': '18'})
plt.rcParams.update({'font.weight': 'bold'})
plt.rcParams.update({'axes.labelweight': 'bold'})
plt.rcParams.update({'axes.titleweight': 'bold'})
```

```python
adata_tnkilc = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_tnkilc_annotated_251017.h5ad')
adata_tnkilc
```

```python
adata_tnkilc.X[0:9,0:9].todense()
```

```python
adata_tnkilc.layers['counts'] = adata_tnkilc.X.copy()  # preserve raw counts

sc.pp.normalize_total(adata_tnkilc, target_sum=1e4)
sc.pp.log1p(adata_tnkilc)
adata_tnkilc
```

```python
adata_tfh = adata_tnkilc[(adata_tnkilc.obs.cellstate.isin(['PD1Lo Tfh', 'PD1Hi Tfh', 'Prlf. Tfh']))]
adata_tfh
```

```python
sc.pl.umap(adata_tfh[sample(adata_tfh.obs.index.tolist(), adata_tfh.n_obs)], color='cellstate', 
           title='Cell State', size = 3)
```

```python
adata_tfh.uns['cellstate_colors'] = ['#E6E6FA', '#A020F0', '#e07b91']
```

```python
sc.pl.umap(adata_tfh[sample(adata_tfh.obs.index.tolist(), adata_tfh.n_obs)], color='cellstate', 
           title='Cell State', size = 3)
```

```python
sc.pp.neighbors(adata_tfh, n_neighbors=20, n_pcs=30, use_rep='X_scVI')
```

```python
sc.tl.umap(adata_tfh, min_dist=0.3, spread=1, random_state=thisseed)
sc.pl.umap(adata_tfh[sample(adata_tfh.obs.index.tolist(), adata_tfh.n_obs)], color='cellstate', 
           title='Cell State', size = 5)
```

```python
tfhtph_markers = [
    'ICOS', 'CCR7', 'MAF', 'IL21',
    'PDCD1', 'CXCL13',  'CXCR5', 'BCL6', 'PRDM1', 
    'CCR5', 'CCR2', 'CX3CR1'
]
```

```python
adata_tfh.obs['Cell State'] = adata_tfh.obs['cellstate']
adata_tfh.uns['Cell State_colors'] = adata_tfh.uns['cellstate_colors']
```

```python
ax = sc.pl.heatmap(
    adata_tfh,
    tfhtph_markers,
    groupby='Cell State',
    swap_axes=True,
    show=False,
    cmap='magma',
    standard_scale='var',
)['groupby_ax'] # Get the group axes object

ax.set_xlabel('')
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
ax.figure.set_size_inches((16,5))
```
