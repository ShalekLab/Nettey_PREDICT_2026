In this notebook, I will create UMAPs for Main and Supplemental figures.

# Setup

```python
import scanpy as sc
import numpy as np
import matplotlib.pyplot as plt

sc.settings.verbosity = 3

thisseed = 231103
np.random.seed(thisseed)

colEPI = "#7FC97F"
colLP = "#FDC086"

sc.set_figure_params(dpi=200, dpi_save=200)
plt.rcParams.update({'font.weight': 'bold'})
plt.rcParams.update({'axes.labelweight': 'bold'})
plt.rcParams.update({'axes.titleweight': 'bold'})
```

# UMAPs

## ColEpi

```python
adata_colepi = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_colepi_annotated_251017.h5ad')
adata_colepi
```

```python
sc.pl.umap(adata_colepi, color='cellsubtype', title='Colonic Epithelium\nCell Subtypes', legend_loc='none')
```

```python
sc.pl.umap(adata_colepi, color='cellsubtype', title='Colonic Epithelium\nCell Subtypes')
```

```python
sc.pl.umap(adata_colepi, color='layer', palette = [colEPI, colLP], 
               title='Colonic Epithelium\nTissue Layer', show=False, add_outline=True)
plt.legend(loc='upper right')
```
