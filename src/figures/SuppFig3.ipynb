This notebook will create the UMAPs in Supplemental Figure 3

# Setup

```python
import scanpy as sc
import numpy as np
import matplotlib.pyplot as plt

sc.settings.verbosity = 3

thisseed = 231103
np.random.seed(thisseed)

colFGID = "#1E88E5"
colICD = "#55D276"
colCCD = "#FFC107"
colUC = "#D81B60"

colCG = {'colepi':"#E41A1C", 'stroma':"#377EB8", 'myeloid':"#4DAF4A",
        'plasma':"#984EA3", 'bcell':"#FF7F00", 'tnkilc':"#A65628"}

framewidth = 4
```

# DiagSub UMAP

```python
sc.set_figure_params(dpi=100)

plt.rcParams.update({'font.weight': 'bold'})
plt.rcParams.update({'axes.labelweight': 'bold'})
plt.rcParams.update({'axes.titleweight': 'bold'})
```

```python
adata_colepi = sc.read_h5ad('../../predict/data/objects/annotated/predictCOL__CB_colepi_annotated_251017.h5ad')
adata_colepi
```

```python
ax = sc.pl.umap(adata_colepi[adata_colepi.obs.timepoint.isin(["T0"])].copy(), 
           color='diagSub', title='Colonic Epithelial Cells', show=False)

ax.spines['top'].set_color(colCG['colepi'])
ax.spines['top'].set_linewidth(framewidth)

ax.spines['right'].set_color(colCG['colepi'])
ax.spines['right'].set_linewidth(framewidth)

ax.spines['bottom'].set_color(colCG['colepi'])
ax.spines['bottom'].set_linewidth(framewidth)

ax.spines['left'].set_color(colCG['colepi'])
ax.spines['left'].set_linewidth(framewidth)

plt.legend(loc = 'best')
plt.show()
```
